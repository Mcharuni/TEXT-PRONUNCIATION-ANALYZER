```python
# 1. Define Target Text and User Input Text
target_text = "The quick brown fox jumps over the lazy dog."
user_text = input(f"Target Text: '{target_text}'\nType your spoken version here: ")

# 2. Convert Graphemes to Phonemes (G2P)
target_phones = [p for p in g2p(target_text) if p != ' ']
user_phones = [p for p in g2p(user_text) if p != ' ']

# 3. Calculate Phonetic Edit Distance & Accuracy Score
distance = Levenshtein.distance(target_phones, user_phones)
max_len = max(len(target_phones), len(user_phones))
accuracy_score = max(0, round((1 - (distance / max_len)) * 100, 2)) if max_len > 0 else 0

# 4. Word-Level Analysis
target_words = target_text.lower().replace(".", "").replace(",", "").split()
user_words = user_text.lower().replace(".", "").replace(",", "").split()

matched_words = [w for w in target_words if w in user_words]
missing_words = [w for w in target_words if w not in user_words]
extra_words = [w for w in user_words if w not in target_words]

# 5. Display Results Report
print("\n" + "="*45)
print("          TEXT PRONUNCIATION REPORT          ")
print("="*45)
print(f"Target Sentence:  {target_text}")
print(f"Your Input:       {user_text}")
print("-" * 45)
print(f"Phonetic Accuracy Score: {accuracy_score}%")
print(f"Correct Words:           {', '.join(matched_words) if matched_words else 'None'}")
print(f"Missing/Mispronounced:   {', '.join(missing_words) if missing_words else 'None'}")
print(f"Unexpected/Extra Words:  {', '.join(extra_words) if extra_words else 'None'}")
print("="*45)
```
